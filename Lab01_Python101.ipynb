# Exercise 2 — Better loops
Unlike many other programming languages, Python lets you iterate directly **over the objects**, instead of iterating over numbers.

This way, on each iteration of the loop the variable holds the value of the list at that position.

````python
car_count = 0

# We iterate over the elements directly, without using an "i" counter
for mode in transport_modes:
    if mode == "Car":
        car_count += 1

print("Cars:", car_count)

# TODO: modify the loop above so it counts EVERY transport mode, and tells you:
# A) Which vehicle is the most used
max_vehicle = ""
for i in transport_modes:
  if transport_modes.count(i) > transport_modes.count(max_vehicle):
    max_vehicle = i
print("Most used:", max_vehicle)

# B) Which one is the least used
min_vehicle = max_vehicle
for i in transport_modes:
  if transport_modes.count(i) < transport_modes.count(min_vehicle):
    min_vehicle = i
print("Least used:", min_vehicle)

# C) Counts - I use a dictionary
count = {}
for i in transport_modes:
  if i in count:
    count[i] += 1
  else:
    count[i] = 1
print(count)
````

# Loops — summary

In Python, the rule of thumb is simple:

*   Iterate over **elements** when you only need to read them:

```python
for mode in transport_modes:      # iterating over elements directly
    if mode == "Car":
        car_count += 1
```

*   Iterate over **indexes** when you need to modify the list:

```python
for i in range(len(prices)):      # iterating over positions 0, 1, 2, ...
    prices[i] = prices[i] * 2     # so we can overwrite each element
```

# Exercise 6 — The Pythonic way: list comprehensions
Python's signature move — building a whole list in **one readable line**:

```python
squares = [n**2 for n in range(1, 6)]         # [1, 4, 9, 16, 25]
evens   = [n for n in range(10) if n % 2 == 0] # with a condition
```

You will see this syntax constantly in real deep learning code.

````python
# TODO 1: solve Exercise 1 again — in ONE line
# Hint: sum(1 for ... in ... if ...)  or  len([... for ... in ... if ...])
n_cars = sum(1 for i in transport_modes if i == "Car")

# TODO 2: solve Exercise 3's " Station" task again — in ONE line
# Re-initialize example_list to its original state to avoid double ' Station'
example_list = ["East Croydon", "Clapham Junction", "Norbury", "Preston Park", "Balcombe"]
stations = [i + " Station" for i in example_list]

print(n_cars)
print(stations)

# Compare with your loop versions: same result, one line. That's Python.
````
