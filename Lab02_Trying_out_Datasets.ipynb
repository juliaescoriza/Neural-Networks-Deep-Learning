```python
# Import IRIS DATASET (FLOWERS)
from sklearn.datasets import load_iris

# Plot library
import matplotlib.pyplot as plt
import seaborn as sns

import pandas as pd
```

```python
# Load flower Dataset
data = load_iris()
```

# Step 1: Structuring the data

Since we are using datasets from the *scikit-learn* library, we know in advance that they have a structure similar to a dictionary that makes it easier to navigate the data.

```python
# Descripció del dataset
print(data.DESCR)
```

```python
# Visualitzem les dades d'algunes mostres
print("\nTarget names(3): \n", data.target_names)
print("\nFeature names (4): \n", data.feature_names)
```

```python
# EXERCISE 2:
# Show the first 5 samples and their classes

print(data.data[:5])
```

```python
# Scatter Plot
_, ax = plt.subplots()
scatter = ax.scatter(data.data[:, 0], data.data[:, 1], c=data.target)
ax.set(xlabel=data.feature_names[0], ylabel=data.feature_names[1])
_ = ax.legend(scatter.legend_elements()[0], data.target_names, loc="lower right", title="Classes")
```

```python
# Function that converts the Dataset into a Pandas Dataframe: https://www.w3schools.com/python/pandas/pandas_dataframes.asp
def sklearn_to_df(sklearn_dataset):
    df = pd.DataFrame(sklearn_dataset.data, columns=sklearn_dataset.feature_names)
    df['target'] = pd.Series(sklearn_dataset.target)
    df['target_names'] = pd.Series(sklearn_dataset.target_names[sklearn_dataset.target])
    return df
```

```python
irisDataset = sklearn_to_df(data)
```

# Step 2: Getting to know the Data

In this dataset we know that we have 4 variables and a target class called **target**.

```python
# Box Plot
plt.figure(figsize = (15, 10))
plt.title("Iris Dataset box plots")
irisDataset.boxplot(grid=False, rot=0, fontsize=13)
plt.show()
```

```python
# EXERCISE 2: Analyze the Box Plot and conclude:

# a) What are the black dots in the sepal width (cm)
# Outliers

# b) What are the green lines?
# Median (or second quartile)

# c) Which independent variable is the most dispersed?
# Petal length (cm)
```

```python
# EXERCISE 3: Locate all the "setosa" flowers and calculate their average petal width

setosa_flowers = irisDataset[irisDataset['target_names'] == 'setosa']
average_petal_width_setosa = setosa_flowers['petal width (cm)'].mean()

print(f"Average petal width for Setosa flowers: {average_petal_width_setosa:.2f} cm")
```

```python
# Discovering a Histogram
fig = plt.figure(figsize = (15,10))
ax = fig.gca()
irisDataset.hist(ax = ax)
plt.show()
```

```python
# Correlation Matrix
plt.figure(figsize=(8, 6))
sns.heatmap(irisDataset.drop(['target', 'target_names'], axis=1).corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```

```python
# EXERCISE 4: Analyse the correlation matrix

# If the coefficient approaches 1, it means that two columns have a strong interrelationship (when one increases, so does the other)
# If the coefficient approaches -1, it means that two columns have a strong interrelationship (when one increases, the other decreases)
# If it approaches 0, it means that there is no linear relationship between the variables
```
